```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import warnings 
warnings.filterwarnings('ignore')
import seaborn as sns
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
from pathlib import Path

data_dir = Path.cwd().parent / "data" / "raw"
train = pd.read_csv(data_dir / "train.csv")
test = pd.read_csv(data_dir / "test.csv")

```

```python
print("Train", train.shape) #number of rows and columns in train
print("Test", test.shape) #number of rows and columns in test
print(train.head(), "\n\n", train.dtypes) #print the first 5 rows of train and the data types of each column
print(test.head(), "\n\n", test.dtypes) #print the first 5 rows of test and the data types of each column
```

```python
# Convert string into datetime (very important!)
train['DateTime'] = pd.to_datetime(train['DateTime']) #convert DateTime column to datetime format
test['DateTime'] = pd.to_datetime(test['DateTime']) #convert DateTime column to datetime format

print("--------")

#check for missing values
print(train.isnull().sum()) #no missing values in train
print(test.isnull().sum()) #no missing values in test

print("--------")
#check for duplicates
print("Duplicate in train are",train.duplicated().sum()) #number of duplicate rows in train
print("Duplicate in test are", test.duplicated().sum()) #number of duplicate rows in test

```

```python
print(train['Vehicles'].describe()) #summary statistics of the target variable
```

```python
print("Junctions in train:", train['Junction'].nunique()) #number of unique junctions
print("Records per junction in train:", train['Junction'].value_counts()) #number of records per junction
```

```python
#Now we will split the data into training and validation sets based on the DateTime column. We will use the first 80% of the data for training and the remaining 20% for validation.
train['is_train'] = 0 #add a new column to indicate whether the row is in the training set or not
test['is_train'] = 1 #add a new column to indicate whether the row is in the test set or not
```

```python
print(train.head())
print(test.head())
```

```python

processed_dir = Path("../data/processed")
processed_dir.mkdir(parents=True, exist_ok=True)

train.to_csv(processed_dir / "train_processed.csv", index=False)
test.to_csv(processed_dir / "test_processed.csv", index=False)

print("Data loading and preprocessing completed.")
```
